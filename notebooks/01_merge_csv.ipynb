```python
import pandas as pd
import os
```

```python
# Thư mục chứa các file cần gộp
input_folder = "../data/processed/"

# Danh sách file bạn muốn gộp
files = [
    "capture104-attackfree.csv",
    "capture104-dosattack.csv",
    "capture104-floodattack.csv",
    "capture104-fuzzyattack.csv",
    "capture104-iec104starvationattack.csv",
    "capture104-mitmattack.csv",
    "capture104-ntpddosattack.csv",
    "capture104-portscanattack.csv",
]

df_list = []
```

```python
# Đọc từng file và đưa vào list
for f in files:
    file_path = os.path.join(input_folder, f)
    print("Đang đọc:", file_path)
    temp = pd.read_csv(file_path)
    df_list.append(temp)

# Gộp thành 1 DataFrame
df_merged = pd.concat(df_list, ignore_index=True)

print("Gộp xong — kích thước cuối cùng:", df_merged.shape)
```

```python
# Lưu file mới
output_file = "../data/processed/iec104_processed.csv"
df_merged.to_csv(output_file, index=False)

print("Đã lưu file gộp tại:", output_file)
```
